```python
### import bibliothèques
import os
from binance.client import Client
```

```python
### connection à binance
API_KEY = os.getenv("OPA_API_KEY")
API_SECRET = os.getenv("OPA_API_SECRET")

client = Client(API_KEY, API_SECRET)
```

```python
def get_data_market(symbol,timeframe,limit) :
    klines = client.get_klines(symbol=symbol,interval=timeframe,limit=limit)
    return klines
```

```python
### exemple
timeframe = Client.KLINE_INTERVAL_1MINUTE
limit = 100
symbol = "BTCEUR"

kline = get_data_market(symbol,timeframe,limit)
```

```python
def generate_signal(klines):
        # attendre qu'on a assez de données
        if len(klines) < 30:
            return "HOLD"

        # Extraire derniers prix
        closes = [float(k[4]) for k in klines]

        # Calcul moyenne mobile courte (10)
        ma_short = sum(closes[-10:]) / 10

        # Calcul moyenne mobile longue (30)
        ma_long = sum(closes[-30:]) / 30

        # Logique de décision
        if ma_short > ma_long:
            return "BUY"

        elif ma_short < ma_long:
            return "SELL"

        return "HOLD"
```

```python
signal = generate_signal(kline)

print("Signal généré :", signal)
```
